```python
edad = int(input("Ingrese su edad: "))
tiene_licencia = input("¿Tiene licencia? (s/n): ")
sancion = input("¿Tiene sanción? (s/n): ")

#Determinar si el usuario puede conducir o no
if edad >= 18 and tiene_licencia =="s" and sancion == "n":
    print("Puede conducir")
else:
    print("No puede conducir")
# ------------------------Problema #2------------------------

for i in range(1, 6):
    print(f"Iteración número {i}")

n = int(input("Escriba una cantidad de numeros: "))
suma = 0
for i in range(1,n+1):
    suma += i
print(suma)

for i in range(1, 6):
    if i % 2 == 0:
        print(i)
#Temperaturas mayores al promedio
# Lista inicial de temperaturas en °C
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
#Promedio de las temperaturas
total =  sum(temperaturas)
cantidad =  len(temperaturas)
promedio = total / cantidad
n = len(temperaturas)
for i in range(n):
    if temperaturas[i] > promedio:
        print(temperaturas[i])

# ------------------------ Problema #3 ------------------------
#Contador hacia atras
contador = 5
while contador > 0:
    print("Contador:", contador)
    contador -= 1

num = int(input("Escriba un numero: "))

while num != 0:
    num = int(input("Escriba un numero: "))
"""Lo representa el ciclo (while), ya que hasta que no se
cumpla la condicion establecida no se termina el ciclo"""

# ------------------------ Problema #4 ------------------------
suma_pares = 0
suma_impares = 0
for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i

sumas = {
    "Suma de los pares" : suma_impares,
    "Suma de los impares": suma_impares,

}
print(sumas)

""" El estado cambia dentro del bucle cuando se le evalua
por medio del condicional if, si cumple la condicion el estado
es uno y si no cumple el condicional el estado cambia """

# ------------------------ Problema #5 ------------------------
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))
    if precio < 0 or cantidad < 0:
        continue
    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]

if productos:
    mas_caro = max(productos, key=lambda p: p["precio"])
    print("Producto más caro:", mas_caro["nombre"], "-", mas_caro["precio"])
else:
    print("No se ingresaron productos válidos.")

print("Total del inventario:", total)
print("Inventario final:", productos)
```
