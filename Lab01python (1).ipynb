```python
explicacion = "La asignación y reasignación cambia según el tipo de objeto, las asignaciones apuntan a elementos en memoria, cuando reasignamos, estamos apuntado a un nuevo objeto en memoria; en Python están los objetos mutables y los inmutables, su diferencia se representa por sus apuntadores, cuando asignamos, apuntaremos a elementos en memoria, con las inmutables, estaremos siempre creando elementos nuevos, mientras que, con las mutables, estaríamos modificando sus atributos pero el objeto general se mantiene, permitiendo modificarlo en lugar de que su referencia apunte a un nuevo elemento en memoria. En este caso al hacer notas = [] lo que hacemos es que la referencia en la función apunte a un nuevo elemento en memoria, para ello, se crea una copia dentro de la función, si quisiéramos que siga la idea original, habría que modificar los elementos o usar métodos que afecten al objeto original."

def agregar_nota(notas, nueva):
    notas.append(nueva)

def reiniciar(notas):
    notas[:] = []
    d = id(notas)
    notas = []
    return id(notas), d

mis_notas = [3.5, 4.0]
idOriginal = id(mis_notas)
agregar_nota(mis_notas, 4.8)
idNuevo, idPrincipal = reiniciar(mis_notas)
print(mis_notas)
print(f"id original: {idOriginal}")
print(f"id al inicio de la función: {idPrincipal}")
print(f"id al final de la función: {idNuevo}")
print(f"id original vs id al inicio de la función ¿iguales? -> {idOriginal == idPrincipal}")
print(f"id original vs id al final de la función ¿iguales? -> {idOriginal == idNuevo}")
print("\n\nMi explicación:\n\n",explicacion)
```

```python
cuenta = {"titular": "Ana", "saldo": 200_000, "movimientos": []}
def consultar(cuenta):
    print(f"\nTitular: {cuenta["titular"]}\nDinero en su cuenta: {cuenta["saldo"]}")
    cuenta["movimientos"].append("Consulta de la cuenta")
    print("Operación exitosa\n")

def depositar(cuenta, monto):
    cuenta["saldo"] += monto
    cuenta["movimientos"].append(f"Deposito de {monto}")
    print("\nOperación exitosa\n")

def retirar(cuenta, monto):
    if cuenta["saldo"] < monto:
        print("\nSaldo insuficiente, operación fallida\n")
    else:
        cuenta["saldo"] -= monto
        cuenta["movimientos"].append(f"Retiro de {monto}")
        print("\nOperación exitosa\n")


accion = input("¿Qué desea hacer hoy? Seleccione escribiendo el número respectivo:\n1) Consultar cuenta\n2) Depostiar en la cuenta\n3) Retirar de la cuenta\n4) Salir\n")

while accion != "4":
    if accion == "1":
        consultar(cuenta)
    elif accion == "2":
        monto = int(input("\nEspecifique el monto a depositar: "))
        depositar(cuenta, monto)
    elif accion == "3":
        monto = int(input("\nEspecifique el monto a retirar: "))
        retirar(cuenta, monto)
    accion = input("¿Qué desea hacer?\n1) Consultar cuenta\n2) Depostiar en la cuenta\n3) Retirar de la cuenta\n4) Salir\n")
print()
print("Movimientos: ")
print(*cuenta["movimientos"], sep = "\n")
```

```python
import random

def juego_adivinanza():
    numero_secreto = random.randint(1, 100)
    intentos_max = 7
    historial = []

    print("=== ¡Bienvenido al Juego de Adivinanza! ===")
    print("He pensado un número entre 1 y 100. Tienes 7 intentos.\n")

    for intento in range(1, intentos_max + 1):
        try:
            num_usuario = int(input(f"Intento {intento}/{intentos_max} - Ingrese su número: "))
        except ValueError:
            print("Por favor, ingrese un número entero válido.")
            continue

        historial.append(num_usuario)

        if num_usuario == numero_secreto:
            print(f"¡Felicidades! Adivinaste el número {numero_secreto} en {intento} intento(s).\n")
            return True, historial
        elif num_usuario < numero_secreto:
            print("El número secreto es **mayor**.\n")
        else:
            print("El número secreto es **menor**.\n")

    print(f"¡Game Over! El número secreto era {numero_secreto}.\n")
    return False, historial

def ejecutar_partidas():
    ganadas = 0
    perdidas = 0

    while True:
        gano, historial = juego_adivinanza()
        if gano:
            ganadas += 1
        else:
            perdidas += 1

        print("Historial de intentos de esta partida:", historial)
        print(f"Marcador actual -> Ganadas: {ganadas} | Perdidas: {perdidas}\n")

        otra = input("¿Desea jugar otra vez? (s/n): ").strip().lower()
        if otra != 's':
            print("¡Gracias por jugar!")
            break

# Para ejecutar el juego:
ejecutar_partidas()
```

```python
import random

def crear_tablero_inicial(filas=10, columnas=10):
    # Tablero de 10x10 aleatorio con 0 (muerta) y 1 (viva)
    return [[random.choice([0, 1]) for _ in range(columnas)] for _ in range(filas)]

def contar_vecinos_vivos(tablero, f, c):
    filas = len(tablero)
    columnas = len(tablero[0])
    vecinos_vivos = 0

    # Recorrer las 8 celdas vecinas (desplazamientos de -1, 0, 1)
    for df in [-1, 0, 1]:
        for dc in [-1, 0, 1]:
            if df == 0 and dc == 0:
                continue  # Ignorar la celda actual

            nf, nc = f + df, c + dc
            # Cuidar los bordes del tablero
            if 0 <= nf < filas and 0 <= nc < columnas:
                vecinos_vivos += tablero[nf][nc]

    return vecinos_vivos

def calcular_siguiente_generacion(tablero):
    filas = len(tablero)
    columnas = len(tablero[0])
    # Se crea un TABLERO NUEVO para la siguiente generación
    nuevo_tablero = [[0 for _ in range(columnas)] for _ in range(filas)]

    for f in range(filas):
        for c in range(columnas):
            vivos = contar_vecinos_vivos(tablero, f, c)
            estado_actual = tablero[f][c]

            # Reglas del Juego de la Vida
            if estado_actual == 1:
                if vivos in [2, 3]:
                    nuevo_tablero[f][c] = 1  # Sobrevive
                else:
                    nuevo_tablero[f][c] = 0  # Muere por soledad o sobrepoblación
            else:
                if vivos == 3:
                    nuevo_tablero[f][c] = 1  # Nace
                else:
                    nuevo_tablero[f][c] = 0

    return nuevo_tablero

def mostrar_tablero(tablero):
    for fila in tablero:
        linea = "".join(["■ " if celda == 1 else "· " for celda in fila])
        print(linea)

# --- Ejecución del simulador para 5 generaciones ---
tablero = crear_tablero_inicial()

print("Generación 0 (Inicial):")
mostrar_tablero(tablero)

for gen in range(1, 6):
    tablero = calcular_siguiente_generacion(tablero)
    print(f"\nGeneración {gen}:")
    mostrar_tablero(tablero)
```

```python
import json

# ==========================================
# 1. VALIDACIÓN DE ENTRADAS CON TRY/EXCEPT/ELSE/FINALLY
# ==========================================

def leer_numero_positivo(mensaje, es_flotante=False):
    while True:
        try:
            entrada = input(mensaje)
            if es_flotante:
                valor = float(entrada)
            else:
                valor = int(entrada)

            if valor < 0:
                print("Error: El valor no puede ser negativo.")
                continue

        except ValueError:
            print("Error: Entrada inválida. Ingrese un valor numérico correcto.")
        else:
            # Se ejecuta SOLO si el bloque try NO lanzó ninguna excepción
            return valor
        finally:
            # Se ejecuta SIEMPRE, haya ocurrido o no una excepción
            pass

def agregar_producto(inventario):
    nombre = input("Ingrese el nombre del producto: ").strip()
    precio = leer_numero_positivo("Ingrese el precio: ", es_flotante=True)
    cantidad = leer_numero_positivo("Ingrese la cantidad: ", es_flotante=False)

    producto = {"nombre": nombre, "precio": precio, "cantidad": cantidad}
    inventario.append(producto)
    print(f"Producto '{nombre}' agregado exitosamente.\n")

# ==========================================
# 2. PERSISTENCIA CON ARCHIVO CSV
# ==========================================

def guardar_inventario_csv(inventario, nombre_archivo="inventario.csv"):
    try:
        with open(nombre_archivo, "w", encoding="utf-8") as f:
            f.write("nombre,precio,cantidad\n")  # Encabezado
            for p in inventario:
                f.write(f"{p['nombre']},{p['precio']},{p['cantidad']}\n")
        print(f"Inventario guardado exitosamente en {nombre_archivo}.")
    except Exception as e:
        print(f"Error al guardar en CSV: {e}")

def cargar_inventario_csv(nombre_archivo="inventario.csv"):
    inventario = []
    try:
        with open(nombre_archivo, "r", encoding="utf-8") as f:
            lineas = f.readlines()
            for linea in lineas[1:]:  # Saltar encabezado
                linea = linea.strip()
                if linea:
                    nombre, precio, cantidad = linea.split(",")
                    inventario.append({
                        "nombre": nombre,
                        "precio": float(precio),
                        "cantidad": int(cantidad)
                    })
        print(f"Inventario cargado exitosamente desde {nombre_archivo}.")
    except FileNotFoundError:
        print(f"El archivo '{nombre_archivo}' no existe. Iniciando con inventario vacío.")
    return inventario

# ==========================================
# 3. PERSISTENCIA CON ARCHIVO JSON
# ==========================================

def guardar_inventario_json(inventario, nombre_archivo="inventario.json"):
    try:
        with open(nombre_archivo, "w", encoding="utf-8") as f:
            json.dump(inventario, f, indent=4, ensure_ascii=False)
        print(f"Inventario guardado exitosamente en {nombre_archivo}.")
    except Exception as e:
        print(f"Error al guardar en JSON: {e}")

def cargar_inventario_json(nombre_archivo="inventario.json"):
    try:
        with open(nombre_archivo, "r", encoding="utf-8") as f:
            inventario = json.load(f)
        print(f"Inventario cargado exitosamente desde {nombre_archivo}.")
        return inventario
    except FileNotFoundError:
        print(f"El archivo '{nombre_archivo}' no existe. Iniciando con inventario vacío.")
        return []

# ==========================================
# PRUEBA / DEMOSTRACIÓN DE FUNCIONAMIENTO
# ==========================================

# Cargar inventario existente si hay, o iniciar vacío
inventario = cargar_inventario_csv()

# Agregar un producto de prueba
agregar_producto(inventario)

# Guardar en CSV y JSON
guardar_inventario_csv(inventario)
guardar_inventario_json(inventario)

# Cargar desde JSON para verificar reconstrucción de datos
inventario_json = cargar_inventario_json()
print("Inventario reconstruido desde JSON:", inventario_json)
```
