```python
import requests

# Llamado a una API pública
url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
usuarios = response.json()
# Procesamiento imperativo
nombres = []
for u in usuarios:
    if u['email'][-4:] == ".biz":
        nombres.append(u["name"])

print("Usuarios encontrados:")
for n in nombres:
    print("-", n)
```

```python
import requests

# Entrada de datos
nombre_pokemon = input("Ingrese el nombre de un Pokémon: ").lower()

# Llamado a la API
url = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"
respuesta = requests.get(url)

# Verificación del estado de la respuesta
if respuesta.status_code == 200:
    datos = respuesta.json()

    # Modelado del estado del Pokémon
    pokemon = {
        "nombre": datos["name"],
        "altura": datos["height"],
        "peso": datos["weight"],
        "experience" : datos['base_experience']
    }

    tipos = []
    for t in datos["types"]:
        tipos.append(t["type"]["name"])
    pokemon["tipos"] = tipos
    habilidades = []
    for h in datos["abilities"]:
        habilidades.append(h["ability"]["name"])
    pokemon["habilidades"] = habilidades


    # Salida del estado actual
    print("\n📘 Información del Pokémon")
    print(f"Nombre: {pokemon['nombre'].capitalize()}")
    print(f"Altura: {pokemon['altura']}")
    print(f"Peso: {pokemon['peso']}")
    print(f"Tipos: {', '.join(pokemon['tipos'])}")
    print(f"Habilidades: {', '.join(pokemon['habilidades'])}")
    print(f"Experiencia base: {pokemon["experience"]}")

else:
    print("Pokémon no encontrado. Verifique el nombre ingresado.")
```

```python
explicacion = "La asignación y reasignación cambia según el tipo de objeto, las asignaciones apuntan a elementos en memoria, cuando reasignamos, estamos apuntado a un nuevo objeto en memoria; en Python están los objetos mutables y los inmutables, su diferencia se representa por sus apuntadores, cuando asignamos, apuntaremos a elementos en memoria, con las inmutables, estaremos siempre creando elementos nuevos, mientras que, con las mutables, estaríamos modificando sus atributos pero el objeto general se mantiene, permitiendo modificarlo en lugar de que su referencia apunte a un nuevo elemento en memoria. En este caso al hacer notas = [] lo que hacemos es que la referencia en la función apunte a un nuevo elemento en memoria, para ello, se crea una copia dentro de la función, si quisiéramos que siga la idea original, habría que modificar los elementos o usar métodos que afecten al objeto original."

def agregar_nota(notas, nueva):
    notas.append(nueva)

def reiniciar(notas):
    notas[:] = []
    d = id(notas)
    notas = []
    return id(notas), d

mis_notas = [3.5, 4.0]
idOriginal = id(mis_notas)
agregar_nota(mis_notas, 4.8)
idNuevo, idPrincipal = reiniciar(mis_notas)
print(mis_notas)
print(f"id original: {idOriginal}")
print(f"id al inicio de la función: {idPrincipal}")
print(f"id al final de la función: {idNuevo}")
print(f"id original vs id al inicio de la función ¿iguales? -> {idOriginal == idPrincipal}")
print(f"id original vs id al final de la función ¿iguales? -> {idOriginal == idNuevo}")
print("\n\nMi explicación:\n\n",explicacion)
```

```python
cuenta = {"titular": "Ana", "saldo": 200_000, "movimientos": []}
def consultar(cuenta):
    print(f"\nTitular: {cuenta["titular"]}\nDinero en su cuenta: {cuenta["saldo"]}")
    cuenta["movimientos"].append("Consulta de la cuenta")
    print("Operación exitosa\n")

def depositar(cuenta, monto):
    cuenta["saldo"] += monto
    cuenta["movimientos"].append(f"Deposito de {monto}")
    print("\nOperación exitosa\n")

def retirar(cuenta, monto):
    if cuenta["saldo"] < monto:
        print("\nSaldo insuficiente, operación fallida\n")
    else:
        cuenta["saldo"] -= monto
        cuenta["movimientos"].append(f"Retiro de {monto}")
        print("\nOperación exitosa\n")


accion = input("¿Qué desea hacer hoy? Seleccione escribiendo el número respectivo:\n1) Consultar cuenta\n2) Depostiar en la cuenta\n3) Retirar de la cuenta\n4) Salir\n")

while accion != "4":
    if accion == "1":
        consultar(cuenta)
    elif accion == "2":
        monto = int(input("\nEspecifique el monto a depositar: "))
        depositar(cuenta, monto)
    elif accion == "3":
        monto = int(input("\nEspecifique el monto a retirar: "))
        retirar(cuenta, monto)
    accion = input("¿Qué desea hacer?\n1) Consultar cuenta\n2) Depostiar en la cuenta\n3) Retirar de la cuenta\n4) Salir\n")
print()
print("Movimientos: ")
print(*cuenta["movimientos"], sep = "\n")
```
